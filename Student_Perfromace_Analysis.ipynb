```python

students = ["Aman", "Riya", "Rahul", "Priya", "Neha"]
marks = [65, 85, 72, 90, 78]

print(students)
print(marks)
```

```python

import pandas as pd

data = {
    "Student": students,
    "Marks": marks
}

df = pd.DataFrame(data)

print(df)
```

```python

print(df.head())

print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())
```

```python

import matplotlib.pyplot as plt

plt.bar(students, marks)

plt.title("Student Marks Comparison")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.show()
```

```python

average_marks = df["Marks"].mean()

plt.bar(["Class Average"], [average_marks], color="orange")

plt.title("Average Student Marks")
plt.ylabel("Marks")

plt.show()

print("Class average:", average_marks)
```

```python

import pandas as pd

df = pd.read_csv("StudentsPerformance.csv")

print(df.head())
```

```python

print("Dataset shape:", df.shape)
```

```python

print(df.columns)
```

```python

df.info()

print("Missing values:")
print(df.isnull().sum())
```

```python

score_columns = [
    "math score",
    "reading score",
    "writing score"
]

print(df[score_columns].describe())
```

```python

average_scores = df[score_columns].mean()

print(average_scores)
```

```python

import matplotlib.pyplot as plt

average_scores.plot(
    kind="bar",
    color=["royalblue", "orange", "seagreen"]
)

plt.title("Average Scores by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

prep_comparison = df.groupby(
    "test preparation course"
)[score_columns].mean()

print(prep_comparison)
```

```python

prep_comparison.plot(kind="bar", figsize=(8, 5))

plt.title("Average Scores by Test Preparation Status")
plt.xlabel("Test Preparation Course")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python

prep_scores = df.groupby("test preparation course")[
    ["math score", "reading score", "writing score"]
].mean()

print(prep_scores)
```

```python

print("STUDENT PERFORMANCE ANALYSIS")
print("-" * 35)

print("Total students:", len(df))

score_columns = ["math score", "reading score", "writing score"]

print("\nAverage scores:")
print(df[score_columns].mean().round(2))

print("\nHighest average subject:")
print(df[score_columns].mean().idxmax())

print("\nLowest average subject:")
print(df[score_columns].mean().idxmin())
```

```python

print("Duplicate rows:", df.duplicated().sum())

print("\nMissing values:")
print(df.isnull().sum())

print("\nScore ranges:")
print(df[["math score", "reading score", "writing score"]].describe())
```

```python

lunch_scores = df.groupby("lunch")[
    ["math score", "reading score", "writing score"]
].mean()

print(lunch_scores.round(2))
```

```python

lunch_scores.plot(kind="bar", figsize=(8, 5))

plt.title("Average Scores by Lunch Type")
plt.xlabel("Lunch Type")
plt.ylabel("Average Score")
plt.xticks(rotation=0)
plt.legend(title="Subject")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df["math score"], bins=10, color="cornflowerblue", edgecolor="black")

plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.tight_layout()
plt.show()
```

```python

score_columns = [
    "math score",
    "reading score",
    "writing score"
]

print("1. Dataset size:", df.shape)
print("\n2. Average scores:")
print(df[score_columns].mean().round(2))

print("\n3. Highest and lowest average:")
averages = df[score_columns].mean()
print("Highest:", averages.idxmax(), round(averages.max(), 2))
print("Lowest:", averages.idxmin(), round(averages.min(), 2))

print("\n4. Average scores by lunch type:")
print(df.groupby("lunch")[score_columns].mean().round(2))

print("\n5. Data quality:")
print("Duplicate rows:", df.duplicated().sum())
print("Missing values:", df.isnull().sum().sum())
```

```python

lunch_avg = df.groupby("lunch")[score_columns].mean()

print(lunch_avg.round(2))

print("\nDifference between lunch groups:")
print(
    (lunch_avg.loc["standard"] - lunch_avg.loc["free/reduced"])
    .round(2)
)
```
